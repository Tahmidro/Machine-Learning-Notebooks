```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
df=load_breast_cancer(as_frame=True).frame
df
```

```python
X=df.drop('target',axis=1)
y=df['target']
```

```python
scaler=StandardScaler()
Xscaled=scaler.fit_transform(X)
```

```python
np.set_printoptions(suppress=True)
pcaFull=PCA(n_components=None)
XpcaFull=pcaFull.fit_transform(Xscaled)

explained=pcaFull.explained_variance_ratio_

explained
```

```python
cumVariance=np.cumsum(explained)
cumVariance
```

```python
plt.plot(cumVariance,marker="o")
plt.xlabel("number of variance")
plt.ylabel("Cummalaitive variance")
plt.grid()
plt.show()
```
